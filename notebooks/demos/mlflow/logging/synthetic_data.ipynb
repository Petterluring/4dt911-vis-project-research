# Synthetic data generation
This notebook generates the synthetic data used by `model_log.ipynb`. The generated data is stored in the `demo_db` database on the remote PostgreSQL server, from which `model_log.ipynb` retrieves the data.


```python
import numpy as np
from pandas import DataFrame
```

```python
SEED = 42
SAMPLES = 500
```

```python
# Generate synthetic data following a linear model: y = ax + b
X = np.linspace(0, 10, SAMPLES)
a = 2
b = 5
y = a * X + b

# Add some noise to the data
np.random.seed(SEED)
mean = 0
std = 2
noise = np.random.normal(mean, std, X.shape)
y_noisy = y + noise

df = DataFrame({"feature": X, "target": y_noisy})
```

```python
# Save data in csv format
df.to_csv("synthetic_data.csv", index=False)
```
