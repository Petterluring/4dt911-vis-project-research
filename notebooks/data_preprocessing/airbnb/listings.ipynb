## Load the File



```python
from pathlib import Path
import json
import pandas as pd

from utils.constants import DATA_PATH

AIRBNB_DATA_PATH = DATA_PATH / "airbnb"
AIRBNB_PROCESSED_DATA_PATH = AIRBNB_DATA_PATH / "processed"

FILE_PATH = AIRBNB_DATA_PATH / "listings.csv"

df = pd.read_csv(FILE_PATH, dtype="string", keep_default_na=False)
original_rows = len(df)
print(f"Rows: {len(df):,} | Columns: {df.shape[1]}")
display(df.head())
```

## 1. Inspect Missing Values and Duplicate IDs

Convert blank fields to missing values. Duplicate listing IDs stop execution for inspection instead of being removed arbitrarily.

```python
df = df.replace(r"^\s*$", pd.NA, regex=True)
display(pd.DataFrame({
    "Missing values": df.isna().sum(),
    "Missing (%)": (df.isna().mean() * 100).round(1),
    "Unique values": df.nunique(),
}))
print("Exact duplicate rows:", int(df.duplicated().sum()))
print("Duplicate listing IDs:", int(df.duplicated("id").sum()))
assert df["id"].notna().all() and not df["id"].duplicated().any()
for column in ["id", "host_id"]:
    assert df[column].str.fullmatch(r"[0-9]+").all(), f"Invalid ID in {column}."
```

## 2. Validate and Convert the Price



```python
has_price = df["price"].notna()
price = pd.to_numeric(df["price"].str.replace(r"[$,]", "", regex=True).str.strip(), errors="coerce")
assert price[has_price].notna().all(), "Inspect unparseable prices."
assert price[has_price].gt(0).all(), "Inspect non-positive prices."

currencies = df.loc[has_price, "price_quote_raw"].map(
    lambda value: json.loads(value)["quote"]["currency"]
)
display(currencies.value_counts().to_frame("Priced listings"))
assert currencies.eq("SEK").all(), "Inspect unexpected currencies."
quote_price = pd.to_numeric(df["price_quote_price_per_night"], errors="raise")
assert quote_price[has_price].notna().all()
assert (price[has_price] - quote_price[has_price]).abs().le(0.011).all(), "Inspect price discrepancies."
df["price"] = price.astype("Float64")
```

### Check Numeric Values Against the Source

Check coordinates, bathrooms and all review scores against the original CSV. The table reports their observed ranges and any changed values. No scaling or decimal-separator substitution is applied.

```python
numeric_source = pd.read_csv(FILE_PATH, dtype="string", keep_default_na=False).replace(r"^\s*$", pd.NA, regex=True)
CHECK_COLUMNS = ["latitude", "longitude", "bathrooms"] + [
    column for column in df.columns if column.startswith("review_scores_")
]
source_numeric = numeric_source.set_index("id")[CHECK_COLUMNS].apply(pd.to_numeric, errors="raise")
source_numeric = source_numeric.loc[df["id"]].reset_index(drop=True)
current_numeric = df[CHECK_COLUMNS].apply(pd.to_numeric, errors="raise").reset_index(drop=True)

numeric_checks = pd.DataFrame({
    "Minimum": current_numeric.min(),
    "Maximum": current_numeric.max(),
    "Changed values": ((current_numeric - source_numeric).abs().gt(1e-10) |
                       current_numeric.isna().ne(source_numeric.isna())).sum(),
})
display(numeric_checks)
assert numeric_checks["Changed values"].eq(0).all()
assert current_numeric["latitude"].between(-90, 90).all()
assert current_numeric["longitude"].between(-180, 180).all()
assert current_numeric["bathrooms"].dropna().ge(0).all()
for column in [c for c in CHECK_COLUMNS if c.startswith("review_scores_")]:
    assert current_numeric[column].dropna().between(0, 5).all()
display(df[["id", "latitude", "longitude", "bathrooms", "bathrooms_text", "review_scores_rating"]].head(10))
```

### Extract the Top 50 Amenities

Rank amenities by the number of **retained, priced listings** that offer them. Count each amenity at most once per listing. Keep the original amenity labels; different labels are not automatically merged. Ties are resolved alphabetically so the top-50 selection is reproducible.

The separate table has two columns: `listing_id` and `amenity`. Each row links one listing to one of the 50 most frequent amenities. `listing_id` joins to `id` in the processed listings file. Amenities outside the top 50 are excluded from this relationship table.

```python
def parse_amenities(value):
    items = json.loads(value)
    assert isinstance(items, list), "Amenities must be a JSON list."
    assert all(isinstance(item, str) for item in items), "Unexpected amenity value."
    return sorted({item.strip() for item in items if item.strip()})

all_listing_amenities = (df[["id", "amenities"]]
    .assign(amenities=lambda data: data["amenities"].map(parse_amenities))
    .explode("amenities")
    .rename(columns={"id": "listing_id", "amenities": "amenity"})
    .dropna(subset=["amenity"])
    .drop_duplicates(["listing_id", "amenity"]))

amenity_counts = (all_listing_amenities.groupby("amenity")["listing_id"]
    .nunique().rename("listing_count").reset_index()
    .sort_values(["listing_count", "amenity"], ascending=[False, True]))
top_50_amenities = amenity_counts.head(50).reset_index(drop=True)
top_50_amenities["listing_percent"] = (
    100 * top_50_amenities["listing_count"] / len(df)
).round(1)
listing_amenities = (all_listing_amenities[
    all_listing_amenities["amenity"].isin(top_50_amenities["amenity"])
].sort_values(["listing_id", "amenity"]).reset_index(drop=True))
listing_amenities["listing_id"] = listing_amenities["listing_id"].astype("string")

assert len(top_50_amenities) == 50
assert listing_amenities["amenity"].nunique() == 50
assert not listing_amenities.duplicated(["listing_id", "amenity"]).any()
assert listing_amenities["listing_id"].isin(df["id"]).all()
display(top_50_amenities)
print("Listing–amenity relationships:", len(listing_amenities))
display(listing_amenities.head(10))
```

### Amenity Frequency Chart

```python
import matplotlib.pyplot as plt

plot_data = top_50_amenities.iloc[::-1]
fig, ax = plt.subplots(figsize=(12, 17))
ax.barh(plot_data["amenity"], plot_data["listing_count"], color="#397fa3")
ax.set_title("Top 50 Amenities in Priced Listings", fontsize=15, pad=14)
ax.set_xlabel("Number of listings")
ax.set_ylabel("Amenity")
ax.set_axisbelow(True)
ax.grid(axis="x", alpha=0.25)
fig.tight_layout()
plt.show()
```

## 3. Keep the Selected Features


```python
KEEP_COLUMNS = ['id', 'host_id', 'name', 'listing_url', 'picture_url', 'latitude', 'longitude', 'property_type', 'room_type', 'accommodates', 'bedrooms', 'beds', 'bathrooms', 'bathrooms_text', 'price', 'minimum_nights', 'maximum_nights', 'availability_30', 'availability_60', 'availability_90', 'availability_365', 'number_of_reviews', 'number_of_reviews_ltm', 'number_of_reviews_l30d', 'first_review', 'last_review', 'reviews_per_month', 'review_scores_rating', 'review_scores_accuracy', 'review_scores_cleanliness', 'review_scores_checkin', 'review_scores_communication', 'review_scores_location', 'review_scores_value']

removed_columns = [column for column in df.columns if column not in KEEP_COLUMNS]
df = df[KEEP_COLUMNS].copy()
print("Removed features:", len(removed_columns))
print("Retained features:", len(df.columns))
```

## 4. Convert Data Types

Convert dates and numeric features explicitly. Keep IDs and descriptive fields as strings. Coordinates, bathrooms and review scores retain decimal points in the CSV; their values are not rescaled.

```python
INTEGER_COLUMNS = ['accommodates', 'bedrooms', 'beds', 'minimum_nights', 'maximum_nights', 'availability_30', 'availability_60', 'availability_90', 'availability_365', 'number_of_reviews', 'number_of_reviews_ltm', 'number_of_reviews_l30d']
FLOAT_COLUMNS = ['latitude', 'longitude', 'bathrooms', 'reviews_per_month', 'review_scores_rating', 'review_scores_accuracy', 'review_scores_cleanliness', 'review_scores_checkin', 'review_scores_communication', 'review_scores_location', 'review_scores_value']
DATE_COLUMNS = ['first_review', 'last_review']

for column in INTEGER_COLUMNS:
    values = pd.to_numeric(df[column], errors="raise")
    assert values.dropna().mod(1).eq(0).all(), f"Non-integer value in {column}."
    df[column] = values.astype("Int64", errors="raise")
for column in FLOAT_COLUMNS:
    df[column] = pd.to_numeric(df[column], errors="raise").astype("Float64")
for column in DATE_COLUMNS:
    assert df[column].dropna().str.fullmatch(r"[0-9]{4}-[0-9]{2}-[0-9]{2}").all()
    df[column] = pd.to_datetime(df[column], format="%Y-%m-%d", errors="raise")
```

## 5. Validate the Result

Check coordinates, stay rules, availability and rating ranges. Missing bedrooms, beds and bathroom counts are left unchanged. Missing review scores can occur for listings without reviews.

```python
assert df["latitude"].between(-90, 90).all()
assert df["longitude"].between(-180, 180).all()
assert df["accommodates"].ge(1).all()
assert df["minimum_nights"].ge(1).all() and df["maximum_nights"].ge(1).all()
assert df["minimum_nights"].le(df["maximum_nights"]).all()
for column in INTEGER_COLUMNS + ["bathrooms", "reviews_per_month"]:
    assert df[column].dropna().ge(0).all(), f"Negative value in {column}."
for days in [30, 60, 90, 365]:
    assert df[f"availability_{days}"].between(0, days).all()
assert (df["availability_30"].le(df["availability_60"]) &
        df["availability_60"].le(df["availability_90"]) &
        df["availability_90"].le(df["availability_365"])).all()
for column in [c for c in df if c.startswith("review_scores_")]:
    assert df[column].dropna().between(0, 5).all(), f"Invalid rating in {column}."
review_dates = df.dropna(subset=["first_review", "last_review"])
assert review_dates["first_review"].le(review_dates["last_review"]).all()
# assert len(df) == original_rows - removed_rows
# assert df["price"].notna().all() and not df["id"].duplicated().any()

display(pd.DataFrame({"Final data type": df.dtypes.astype(str), "Remaining missing values": df.isna().sum()}))
display(df[["price", "accommodates", "bedrooms", "beds", "bathrooms"]].describe())
print(f"Final rows: {len(df):,} | Final columns: {df.shape[1]}")
```

## 6. Add DeSO Codes and Stockholm Districts

### Local Boundary Files

Use `DeSO_2025.gpkg` and `stockholm_districts_2025.geojson`.

https://www.scb.se/en/services/open-data-api/open-geodata/open-data-for-deso--demographic-statistical-areas/
https://services2.arcgis.com/uM75dQp2RlOXWzhF/ArcGIS/rest/services/Stadsdelsn%C3%A4mnder_2025/FeatureServer/7/query


DeSO codes identify small statistical areas. Districts are assigned independently using the 11 district polygons in the GeoJSON.

```python
import geopandas as gpd

DESO_PATH = AIRBNB_DATA_PATH / "DeSO_2025.gpkg"
DISTRICTS_PATH = AIRBNB_DATA_PATH / "stockholm_districts_2025.geojson"

assert DESO_PATH.exists(), f"Missing DeSO file: {DESO_PATH}"
assert DISTRICTS_PATH.exists(), f"Missing district file: {DISTRICTS_PATH}"

# Allow this section to be rerun without duplicate derived columns.
df = df.drop(columns=["municipality_name", "municipality_code", "regso_code", "deso_code", "district"], errors="ignore")
listings_before_geo = df.copy()
assert df["id"].notna().all() and not df["id"].duplicated().any()

coords = df[["longitude", "latitude"]].apply(pd.to_numeric, errors="coerce")
valid = (coords["longitude"].between(-180, 180) & coords["latitude"].between(-90, 90)).fillna(False).astype(bool)
points = gpd.GeoDataFrame(
    df.loc[valid, ["id"]].copy(),
    geometry=gpd.points_from_xy(coords.loc[valid, "longitude"], coords.loc[valid, "latitude"]),
    crs="EPSG:4326",
)
print("Invalid coordinates:", int((~valid).sum()))
```

### Assign the DeSO Code

Load the `DeSO_2025` layer, preserve its `desokod` values as strings and merge polygon parts belonging to the same code. A listing receives a code only when it intersects exactly one DeSO area. Multiple matches at a shared boundary remain unassigned; no nearest-area fallback is used.

```python
deso = gpd.read_file(DESO_PATH, layer="DeSO_2025")
assert "desokod" in deso.columns and deso.crs is not None
assert deso.geometry.notna().all() and not deso.geometry.is_empty.any()
assert deso.geometry.geom_type.isin(["Polygon", "MultiPolygon"]).all()
assert deso.geometry.is_valid.all(), "Inspect invalid DeSO geometries."
deso = deso[["desokod", deso.geometry.name]].rename(columns={"desokod": "deso_code"})
assert deso["deso_code"].notna().all()
deso["deso_code"] = deso["deso_code"].astype("string").str.strip()
assert deso["deso_code"].ne("").all()
deso = deso.dissolve(by="deso_code", as_index=False)
assert deso.geometry.is_valid.all()

deso_joined = gpd.sjoin(points.to_crs(deso.crs), deso, how="left", predicate="intersects")
deso_matches = deso_joined.groupby("id")["deso_code"].nunique()
deso_unique = deso_joined.loc[deso_joined["id"].map(deso_matches).eq(1), ["id", "deso_code"]].drop_duplicates()
assert not deso_unique["id"].duplicated().any()
df["deso_code"] = df["id"].map(deso_unique.set_index("id")["deso_code"]).astype("string")
print("Distinct DeSO areas:", df["deso_code"].nunique())
print("Ambiguous DeSO assignments:", int(deso_matches.gt(1).sum()))
print("Unassigned DeSO codes:", int(df["deso_code"].isna().sum()))
```

### Load and Standardize the 11 Districts

Use the `Namn` field, standardize the district names and dissolve polygon parts after converting to SWEREF 99 TM (EPSG:3006). Stop execution if the file does not contain 11 distinct districts.

```python
districts = gpd.read_file(DISTRICTS_PATH)
assert "Namn" in districts.columns and districts.crs is not None
assert districts["Namn"].notna().all()
assert districts.geometry.notna().all() and not districts.geometry.is_empty.any()
assert districts.geometry.geom_type.isin(["Polygon", "MultiPolygon"]).all()
assert districts.geometry.is_valid.all(), "Inspect invalid district geometries."

name_mapping = {
    "Norra innerstadens": "Norra innerstaden",
    "Södermalms": "Södermalm",
    "Kungsholmens": "Kungsholmen",
    "Skarpnäcks": "Skarpnäck",
    "Skärholmens": "Skärholmen",
    "Enskede-Årsta-Vantörs": "Enskede-Årsta-Vantör",
}
districts["district"] = districts["Namn"].str.strip().replace(name_mapping)
assert districts["district"].ne("").all()
districts = districts[["district", "geometry"]].to_crs("EPSG:3006").dissolve(by="district").reset_index()
assert districts.geometry.is_valid.all()
if len(districts) != 11:
    raise ValueError(f"Expected 11 districts, found {len(districts)}.")
display(districts[["district"]])
```

### Assign Districts and Check the Result

Use `within` to assign points strictly inside district polygons. Points exactly on a boundary or outside the district coverage remain unassigned. If polygons overlap and produce multiple district matches, leave the district empty and report the ambiguity rather than selecting an arbitrary match.

Map results back using the unique listing ID to preserve row order and index. Retain listings with missing assignments for inspection.

```python
joined = gpd.sjoin(points.to_crs(districts.crs), districts[["district", "geometry"]],
                   how="left", predicate="within")
district_matches = joined.groupby("id")["district"].nunique()
district_unique = joined.loc[joined["id"].map(district_matches).eq(1), ["id", "district"]].drop_duplicates()
assert not district_unique["id"].duplicated().any()
df["district"] = df["id"].map(district_unique.set_index("id")["district"]).astype("string")

pd.testing.assert_frame_equal(df[listings_before_geo.columns], listings_before_geo)
print("Ambiguous district assignments:", int(district_matches.gt(1).sum()))
print("Unassigned districts:", int(df["district"].isna().sum()))
display(df["district"].value_counts(dropna=False).to_frame("Listings"))
display(df[["id", "deso_code", "district"]].head(10))
display(df.loc[df["district"].isna() | df["deso_code"].isna(),
               ["id", "latitude", "longitude", "deso_code", "district"]].head(20))
```
