# Stockholm Points of Interest (POI) — Enriched Dataset

## 1. Overview

This dataset contains **4,234 Points of Interest (POIs)** located within Stockholm, Sweden. It was originally collected from [OpenStreetMap (OSM)](https://www.openstreetmap.org/) using [Overpass Turbo](https://overpass-turbo.eu/) and subsequently enriched using Python.

The purpose of the enrichment is to transform raw geographical information into a structured dataset suitable for **spatial analysis, tourism research, and interactive data visualization**.

The dataset is part of the **4DT911 visualization project**, which investigates spatial patterns in Stockholm's Airbnb rental market and how factors such as nearby attractions, restaurants, nightlife, and other facilities relate to Airbnb listing prices and availability.

The enrichment preserves existing OpenStreetMap attributes while introducing standardized classifications, spatial proximity measurements, data completeness indicators, and fields prepared for future integration with external statistics.

### Dataset statistics

| Attribute | Original dataset | Enriched dataset |
|---|---|---|
| Number of locations | 4,234 | 4,234 |
| Number of columns | 438 | 495 |
| Geographic coverage | Stockholm | Stockholm |
| Original data source | OpenStreetMap | OpenStreetMap |
| Coordinates | Latitude / Longitude | Latitude / Longitude |
| Standardized classifications | No | Yes |
| Spatial proximity features | No | Yes |
| Data completeness indicators | No | Yes |
| External statistics | No | Prepared but not populated |

A smaller analysis-oriented version containing **79 columns** is also provided.

## 3. Original Data

The original dataset was exported from OpenStreetMap using an Overpass Turbo query targeting the Stockholm administrative area.

The query selected named establishments and locations from four main OpenStreetMap categories.

| OSM category | Example values |
|---|---|
| `amenity` | restaurant, cafe, bar, pub, nightclub, theatre |
| `tourism` | museum, gallery, attraction, viewpoint, artwork |
| `historic` | monument, memorial, castle, archaeological_site |
| `leisure` | park, garden, marina, water_park |

Each location includes geographical coordinates and any additional attributes available in OpenStreetMap.

The original dataset contains both tourism-related establishments and everyday amenities. Therefore, not every POI should be interpreted as a tourist attraction.

### Data source

- OpenStreetMap: https://www.openstreetmap.org/
- Overpass Turbo: https://overpass-turbo.eu/
- OpenStreetMap documentation: https://wiki.openstreetmap.org/

## 4. Data Cleaning and Validation

Before enrichment, the dataset undergoes several preprocessing operations.

### 4.1 Coordinate validation

Coordinates are converted to numeric values.

Records without valid names or geographical coordinates are excluded.

Valid coordinate ranges are:

- Latitude: -90 to 90 degrees
- Longitude: -180 to 180 degrees

All 4,234 original records passed the required validation checks.

### 4.2 Duplicate handling

Duplicate OpenStreetMap identifiers are removed.

No records were removed during this step.

However, separate OpenStreetMap objects may still represent the same physical establishment. These potential real-world duplicates are not automatically merged.

### 4.3 Missing values

Missing optional attributes are preserved as missing values.

For example, if a restaurant does not have a recorded website or opening hours, those attributes remain empty.

No missing values are replaced with fabricated information.

## 5. Enrichment Features

The enrichment introduces several groups of new attributes.

### 5.1 Standardized classification

OpenStreetMap allows a location to have several tags across different categories.

The enrichment introduces a consistent classification system using three attributes.

| Attribute | Description |
|---|---|
| `primary_category` | Main OpenStreetMap category |
| `primary_subcategory` | Specific type within the main category |
| `analysis_group` | Higher-level classification for analysis |

The primary category is assigned using the following priority:

1. Tourism
2. Historic
3. Leisure
4. Amenity

The original classification tags are preserved.

For example:

| Name | Primary category | Subcategory | Analysis group |
|---|---|---|---|
| Example museum | tourism | museum | visitor_attraction |
| Example restaurant | amenity | restaurant | food_drink |
| Example nightclub | amenity | nightclub | nightlife |
| Example monument | historic | monument | culture_heritage |

These examples illustrate the classification logic.

### 5.2 Analysis groups

Each POI is assigned to one of the following groups:

| Analysis group | Description | Locations |
|---|---|---:|
| `food_drink` | Restaurants and cafés | 2,397 |
| `culture_heritage` | Artwork, monuments, and historical locations | 653 |
| `parks_recreation` | Parks, gardens, and recreational facilities | 526 |
| `nightlife` | Bars, pubs, nightclubs, and related venues | 321 |
| `visitor_attraction` | Museums, galleries, attractions, and viewpoints | 249 |
| `culture_entertainment` | Cinemas, theatres, and arts centres | 88 |
| **Total** | | **4,234** |

This classification simplifies filtering and comparison between different types of establishments.

The `analysis_group` represents an analytical grouping, not an official OpenStreetMap classification.

## 6. Standardized Facility Information

The original OpenStreetMap export contains multiple fields that may represent the same information.

For example, a website can be stored as `website`, `contact:website`, or `url`.

To simplify analysis, the enrichment creates standardized attributes by selecting the first available value from predefined alternatives.

### New attributes

| Attribute | Description |
|---|---|
| `best_website` | Preferred website URL |
| `best_phone` | Preferred telephone number |
| `best_email` | Preferred email address |
| `best_instagram` | Instagram information |
| `best_facebook` | Facebook information |
| `display_name_en` | English name if available, otherwise original name |

The full enriched dataset contains all these normalized fields. The smaller analysis dataset includes the five contact fields but does not include `display_name_en`.

The original OpenStreetMap values remain unchanged in the full dataset.

### Additional available information

Depending on the establishment, the dataset may also contain:

- Opening hours
- Cuisine types
- Wheelchair accessibility
- Outdoor and indoor seating
- Vegan and vegetarian options
- Entrance fees
- Operator information
- Descriptions
- Postal addresses
- Websites and social media
- Wikipedia and Wikidata identifiers

The availability of these attributes varies considerably between establishments.

## 7. Spatial Enrichment

The most significant addition is the calculation of spatial relationships between locations.

The original dataset identifies where facilities are located.

The enriched dataset additionally describes **how many other facilities are nearby and how far away different categories of establishments are**.

### 7.1 Geographic coordinate system

The original coordinates use:

`EPSG:4326 — WGS84`

For spatial distance calculations, coordinates are projected to:

`EPSG:3006 — SWEREF99 TM`

This allows distances to be calculated in meters.

All measurements represent **straight-line geographic distances**, not walking distances, driving distances, or travel times.

### 7.2 General POI density

For each location, the enrichment calculates the number of other POIs within specified distances.

| Attribute | Description |
|---|---|
| `all_pois_within_250m` | Number of other POIs within 250 meters |
| `all_pois_within_500m` | Number of other POIs within 500 meters |
| `all_pois_within_1000m` | Number of other POIs within 1 kilometer |

The selected POI itself is excluded from these counts.

These attributes provide simple measurements of local establishment density.

For example, a POI in central Stockholm may have a substantially higher nearby POI count than one in a residential neighborhood.

### 7.3 Category-specific POI density

Additional attributes measure how many establishments belonging to a particular analysis group are located nearby.

The following naming convention is used:

`{analysis_group}_within_{distance}m`

Examples:

| Attribute | Description |
|---|---|
| `visitor_attraction_within_500m` | Visitor attractions within 500 m |
| `visitor_attraction_within_1000m` | Visitor attractions within 1 km |
| `food_drink_within_500m` | Restaurants and cafés within 500 m |
| `nightlife_within_500m` | Nightlife establishments within 500 m |
| `culture_heritage_within_1000m` | Cultural and historical sites within 1 km |
| `parks_recreation_within_1000m` | Parks and recreational facilities within 1 km |

Equivalent 500-meter and 1-kilometer features are created for the other analysis groups.

These attributes make it possible to compare different locations according to their surrounding facilities.

### 7.4 Distance to the nearest establishment

The enrichment also calculates the distance to the nearest POI belonging to each analysis group.

The naming convention is:

`nearest_{analysis_group}_m`

Examples:

| Attribute | Description |
|---|---|
| `nearest_visitor_attraction_m` | Distance to nearest visitor attraction |
| `nearest_food_drink_m` | Distance to nearest restaurant or café |
| `nearest_nightlife_m` | Distance to nearest nightlife venue |
| `nearest_culture_heritage_m` | Distance to nearest cultural or historical POI |
| `nearest_parks_recreation_m` | Distance to nearest park or recreational POI |

For locations belonging to the queried category, the calculation excludes the location itself.

When no other POI exists in a particular category, the distance is missing.

### 7.5 Calculation method

Spatial proximity is calculated using:

- `pyproj` for coordinate transformations
- `scipy.spatial.cKDTree` for efficient spatial queries
- NumPy for numerical operations

KD-trees allow nearby establishments to be identified without comparing every possible pair of locations individually.

This makes the enrichment suitable for larger geographic datasets.

| `has_phone` | Whether a telephone number is available |
| `has_email` | Whether an email address is available |
| `has_wikidata` | Whether a Wikidata identifier exists |
| `has_wikipedia` | Whether a Wikipedia reference exists |
| `has_opening_hours` | Whether opening hours are recorded |
| `has_description` | Whether a description is available |

The enrichment also introduces:

`original_osm_fields_present`

This attribute counts how many original OpenStreetMap fields contain non-missing values for a location.

These features are useful for data exploration, filtering, and identifying incomplete records.

They should not be interpreted as measurements of establishment quality, popularity, or tourism demand.

## 13. Limitations

Several limitations should be considered when analyzing the dataset.

### Incomplete OpenStreetMap coverage

OpenStreetMap is collaboratively maintained, and not every establishment in Stockholm is necessarily represented.

Data completeness may vary between neighborhoods and categories.

### POI counts do not equal tourist volume

The presence of tourist attractions or restaurants does not directly measure how many tourists visit a location.

A small café and a major museum may each contribute one POI to a density calculation, despite receiving very different numbers of visitors.

### Missing optional attributes

Many locations lack websites, opening hours, contact details, or accessibility information.

Missing values do not necessarily indicate that a facility lacks the corresponding service.

### Straight-line distance

The calculated distances do not account for road networks, waterways, bridges, or pedestrian accessibility.

Real walking distances may be substantially longer.

### Potential physical duplicates

Duplicate OpenStreetMap identifiers are removed, but multiple OSM elements may still describe the same real-world establishment.

### Temporal limitations

The dataset represents OpenStreetMap information available at the time of extraction.

It should not automatically be treated as an accurate description of establishments that existed during earlier Airbnb observation periods.

### Geographic representation

The CSV primarily uses point coordinates. The original GeoJSON should be retained when analyzing geographic shapes, such as building footprints or park boundaries.

### External statistics

Google ratings, visitor statistics, and business revenue information have not yet been collected.

Their corresponding columns are currently empty.
